# Resume and Portfolio Packaging
## Project Title
Customer Retention and Churn Analysis for an E-commerce Platform

---

## Business Problem
Customer acquisition costs are high, yet repeat purchase rates are low.  
This project analyzes customer purchasing behavior to understand churn, identify retention gaps, and quantify the revenue impact of low repeat engagement.

---

## Dataset Overview
- Brazilian E-commerce Dataset (Olist)
- Approximately 110,000 orders and 96,000 customers
- Time period: 2016–2018
- Multiple relational tables merged and validated

---

## Tools and Techniques
- Python (Pandas, NumPy)
- Data Visualization (Matplotlib)
- Cohort Analysis
- Customer Segmentation
- Churn Proxy Definition
- Business KPI Design

---

## Analytical Scope
- First-time vs repeat customer segmentation
- Time between purchases analysis
- Monthly cohort-based retention analysis
- Revenue contribution by customer type
- Customer churn estimation using inactivity thresholds

---

## Key Insights
- A large majority of customers churn within 90 days of their first purchase
- Repeat customers represent a small share of total customers and revenue
- Most repeat purchases occur within 30–60 days after the first order
- Cohort analysis shows consistent retention decay across acquisition periods
- Revenue growth is driven primarily by first-time purchases rather than customer lifetime value

---

## Business Impact
- Identified customer retention as the primary growth bottleneck
- Quantified revenue inefficiency caused by high churn rates
- Highlighted the critical post-purchase engagement window
- Provided a data-backed foundation for retention-focused strategies

---

## Strategic Recommendations
- Implement lifecycle-based communication within 30–60 days post-purchase
- Introduce incentives for second purchases to improve early retention
- Track cohort retention and repeat purchase rate as core performance metrics
- Shift success measurement from acquisition volume to customer lifetime value (CLV)
