## Executive Summary
Customer acquisition is costly, yet the majority of customers do not return after their first purchase. 
Through retention, cohort, and churn analysis, we identify rapid post-purchase disengagement as the primary driver of low repeat purchase and suboptimal revenue efficiency.

This notebook synthesizes analytical findings into business insights and actionable strategies.

---

## Key Business Insights

### 1️⃣ Repeat Purchase Is Exceptionally Low
- Over **80% of customers churn** within 90 days of their first purchase.
- The customer base is overwhelmingly composed of first-time buyers.
- Repeat customers represent a **minority of total users**.

📌 **Business Impact:**  
The company relies heavily on continuous customer acquisition rather than sustainable retention.

---

### 2️⃣ Customer Engagement Drops Quickly After First Order
- Most repeat purchases happen within **30–60 days**.
- After 90 days, the likelihood of return drops sharply.
- Time-between-purchase distribution shows a steep right-skew.

📌 **Business Impact:**  
There is a narrow opportunity window to influence repeat behavior.

---

### 3️⃣ Cohort Analysis Reveals Structural Retention Issues
- All cohorts show sharp drop-off after month 0.
- No cohort demonstrates improving retention over time.
- Retention decay is consistent across acquisition periods.

📌 **Business Impact:**  
Retention issues are systemic and not caused by seasonality or cohort quality.

---

### 4️⃣ Revenue Is Dominated by First-Time Purchases
- First-time buyers generate the majority of revenue.
- Repeat customers contribute disproportionately less revenue.
- High lifetime value potential remains untapped.

📌 **Business Impact:**  
The business is missing a high-margin growth lever: repeat customer monetization.

---

## Root Cause Diagnosis

| Observation | Root Cause |
|------------|-----------|
| High churn | Lack of post-purchase engagement |
| Low repeat rate | No incentive for second purchase |
| Fast disengagement | Absence of lifecycle-based communication |
| Revenue inefficiency | Overdependence on paid acquisition |

---

## Strategic Recommendations

### 🎯 Short-Term (0–3 Months)
- Launch **30–60 day reactivation campaigns**
- Offer **second-purchase discounts**
- Trigger automated follow-ups after order delivery

---

### 🧠 Medium-Term (3–6 Months)
- Introduce **customer segmentation** based on purchase timing
- Personalize offers using category affinity
- Deploy loyalty or reward-based incentives

---

### 📈 Long-Term (6–12 Months)
- Shift success metrics from acquisition volume to **Customer Lifetime Value (CLV)**
- Track cohort retention as a core KPI
- Develop churn prediction models for proactive retention

---

## KPIs to Monitor Going Forward

| Metric | Purpose |
|-----|-----|
| Repeat Purchase Rate | Retention health |
| Time Between Purchases | Engagement speed |
| Cohort Retention | Long-term loyalty |
| Churn Rate | Customer loss |
| CLV | Sustainable growth |

---

## Final Recommendation
Improving retention by even a small margin can significantly reduce acquisition dependency and unlock compounding revenue growth. 
The highest ROI opportunity lies in activating customers within the first 60 days post-purchase.

---

## Conclusion
This analysis demonstrates that low repeat purchase is driven by rapid churn rather than poor acquisition quality. 
By focusing on lifecycle engagement and retention-driven strategies, the business can achieve more sustainable and profitable growth.
