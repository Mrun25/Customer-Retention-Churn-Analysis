```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("cleaned_ecommerce_data.csv", parse_dates=[
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date']).copy()
```

```python
df.shape
df.isna().sum()
```

```python
df['order_status'].value_counts(normalize = True) * 100
```

```python
df['revenue'] = df['price'] + df['freight_value']
```

```python
df[['price', 'freight_value', 'revenue']].head(10)
```

```python
total_revenue = df['revenue'].sum()
total_revenue
```

```python
df['order_month'] = df['order_purchase_timestamp'].dt.to_period('M')

# Monthly revenue
monthly_revenue = (df.groupby('order_month')['revenue'].sum().sort_index())

monthly_revenue
```

```python
# Each product revenue
category_revenue = df.groupby('product_category_name')['revenue'].sum().sort_values(ascending = False)
category_revenue
```

```python
# Each product being sold
category_order = df.groupby('product_category_name')['order_id'].nunique().sort_values(ascending = False)
category_order
```

```python
# Each customer generated revenue
customer_revenue = df.groupby('customer_unique_id')['revenue'].sum().sort_values(ascending = False)
customer_revenue
```

```python
orders_per_customer = (df.groupby('customer_unique_id')['order_id'].nunique())

repeat_rate = (orders_per_customer > 1).mean() * 100
repeat_rate
```

```python
df['delivery_delay_day'] = (df['order_delivered_customer_date'] - df['order_estimated_delivery_date']).dt.days
```

```python
(df['delivery_delay_day'] > 0).mean() * 100
```

```python
df['delivery_delay_day'].describe()
```
